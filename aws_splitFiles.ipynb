```python
import pandas as pd
import numpy as np
import re
from tqdm import tqdm

import os
import shutil
```

```python
df=pd.read_csv("AWS_Search_Cust.csv")
df.head(2)
```

```python
for i in tqdm(range(df.shape[0])):
    name=df.iloc[i,df.columns.get_loc("CUST_NAME")]
    row_num=df.iloc[i,df.columns.get_loc("ROW")]
    df_ = pd.DataFrame(data={"CUST_NAME": [name],"ROW":[row_num]},columns=["CUST_NAME","ROW"],index=None)
    df_.to_csv(r"dump\\"+str(row_num)+".csv",index=False)
```

```python
N = 100000  # the number of files in seach subfolder folder


def move_files(abs_dirname):
    """Move files into subdirectories."""

    files = [os.path.join(abs_dirname, f) for f in os.listdir(abs_dirname)]

    i = 0
    curr_subdir = None
    files.sort()

    for f in files:
        # create new subdir if necessary
        if i % N == 0:
            subdir_name = os.path.join(abs_dirname, '{0:03d}'.format(i // N + 1))
            os.mkdir(subdir_name)
            curr_subdir = subdir_name

        # move file to current dir
        f_base = os.path.basename(f)
        shutil.move(f, os.path.join(subdir_name, f_base))
        i += 1
        
```

```python
move_files(r"dump//")
```
