This code combines all files in a folder. All files have to be the same type/format/layout

```python
import pandas as pd
import os
import glob
import re
from tqdm import tqdm
```

```python
path = r'C:\Users\F5385180\OneDrive - FRG\Python_Scripts_Shikash\Webscraping\NPO_Pages' # use your path
all_files = glob.glob(os.path.join(path, "*.csv")) 
lis=[]
for file in tqdm(all_files):
    if file.endswith(".csv"):  
        df=pd.read_csv(file)
        lis.append(df)
df_1= pd.concat(lis, axis=0, ignore_index=True)
df_1.head(3)
```

```python
df_1.to_csv("NPO_List_9May.csv")
```
