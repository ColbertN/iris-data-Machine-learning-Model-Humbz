```python
from bs4 import BeautifulSoup
import pandas as pd
import time
import random
import web_general
import requests
```

```python
url = 'http://www.npo.gov.za/PublicNpo/Npo'
```

```python
def next_page():
    payload = {
        "X-Requested-With": "XMLHttpRequest",
        "Content-Type": "application/x-www-form-urlencoded; charset=UTF-8",
    }
    data = {
        "PublicNpoSearch.IndexGrid.GridSettings.PageSize": "20",
        "PublicNpoSearch.IndexGrid.GridSettings.CurrentPage": "2",
        "PublicNpoSearch.IndexGrid.GridSettings.SortColumn": "",
        "PublicNpoSearch.IndexGrid.GridSettings.SortOrder": "",
    }
    response = requests.post(url, headers=payload, data=data)
    return response

def get_details_save(rows_, count, df):
    for row in rows_[1:]:
        hold_list = []
        for item in row:
            hold_list.append(item.text)
        hold_list.append(count)
        df.loc[len(df)] = hold_list
    return df

def get_details(rows_, count):
    save_list = []
    last_name = ""
    for row in rows_[1:]:
        hold_list = []
        for item in row:
            hold_list.append(item.text)
        hold_list.append(count)
        save_list.append(hold_list)
        last_name = hold_list[4]
    return save_list, last_name

def get_rows(response):
    soup = BeautifulSoup(response.content, 'html.parser')
    table = soup.find(lambda tag: tag.name == 'table' and tag.has_attr('id') and tag['id'] == "PublicNpoSearch_Index-Table")
    rows = table.findAll(lambda tag: tag.name == 'tr')
    page_view_element = soup.find("div", class_="scroll ui-state-default ui-jqgrid-pager ui-corner-bottom")

    if page_view_element:
        page_view = page_view_element.text
    else:
        page_view = "0 of 0"

    if "-" in page_view and "of" in page_view:
        page_num = int(page_view.split("-")[1].split("of")[0].replace(" ", ""))
    else:
        page_num = 0

    page_view = int(page_num / 20)
    return rows, page_view
```

```python
page_num_df = 11324
check = 0
count = 0

while page_num_df <= 12904:
    if check == 0:
        response = requests.get(url)
    else:
        response = next_page()

    rows, page_view = get_rows(response)

    if page_num_df != page_view:
        df = pd.DataFrame(columns=["ID", "Search", "Name", "Province", "Reg_no", "DOR", "Status", "DC", "FM", "Page"])
        save_list, last_name = get_details(rows, count)
        df = get_details_save(rows, page_view, df)
        page_num_df = df.iloc[-1, df.columns.get_loc("Page")]
        print(page_num_df, end="\r")
        page_num_df = page_view

        df.to_csv(f"C:\\Users\\F5448255\\OneDrive - FRG\\Documents\\NPO_Pages\\{page_num_df}.csv", index=False)
        check = 0
    else:
        check += 1

    if check >= 30:
        response = next_page()
        check = 1
```
